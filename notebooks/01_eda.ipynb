### How does each Feature relate to the Target?

#### Numerical Featres vs. Target

##### Does more RAM correspond to higher price? 

```python
# lets start with RAM
# Does more RAM correspond to higher price?

plt.figure(figsize=(10, 6))
plt.scatter(df["RAM (GB)"], df["Price (Euro)"])
plt.xlabel("RAM (GB)")
plt.ylabel("Price (Euro)")
plt.title("RAM vs Laptop Price")
plt.show()
```

Yes. In general, laptops with higher RAM tend to have higher prices. However, this relationship is not true for every laptop, since other features such as CPU, GPU, storage, and brand also influence the price.

##### Does weight relate to price?


```python
# lets see the weight realtion 
# Does weight relate to price?

plt.figure(figsize = (10, 6))
plt.scatter(df["Weight (kg)"], df["Price (Euro)"])
plt.xlabel("Weight (kg)")
plt.ylabel("Price (Euro)")
plt.title("Weight vs Laptop Price")
plt.show()
```

The plot does not show a clear upward or downward trend between laptop weight and price. Therefore, there does not appear to be a strong linear relationship between weight and price in this dataset.

#### Categorical Features vs. Target

##### Do companies have different price distributions?

```python
# Lets start with Company 
# Do companies have different price distributions?

companies = df["Company"].unique()

plt.figure(figsize=(12, 6))

df.boxplot(column="Price (Euro)", by="Company")

plt.xlabel("Company")
plt.ylabel("Price (Euro)")
plt.suptitle("")
plt.title("Laptop Price by Company")

plt.xticks(rotation=45)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(15, 7))

df.boxplot(
    column="Price (Euro)",
    by="Company",
    ax=ax
)

ax.set_xlabel("Company")
ax.set_ylabel("Price (Euro)")
ax.set_title("Laptop Price by Company")

plt.suptitle("")
plt.xticks(rotation=45)

plt.show()
```


Yes. The chart shows that laptop prices have different distributions across companies. For example, Razer laptops tend to have a relatively high price distribution compared with many other brands. However, this does not mean that the company itself causes higher prices, since other features such as RAM, CPU, GPU, and storage can also influence the price.


##### Do different types of laptops have different price distributions?

```python
# Lets compare the next feature which is TypeName
# Do different types of laptops have different price distributions?

fig, ax = plt.subplots(figsize=(12, 6))

df.boxplot(column="Price (Euro)", by= "TypeName", ax = ax)

ax.set_xlabel("Laptop Type")
ax.set_ylabel("Price (Euro)")
ax.set_title("Laptop Price by Type")

plt.suptitle("")
plt.xticks(rotation=30)

plt.show()
```


The box plot shows differences in price distributions across laptop types. For example, Netbooks have the lowest observed minimum price, while Gaming and Workstation laptops have relatively high minimum prices. Ultrabooks and Gaming laptops also have similar minimum prices. There is nevertheless considerable overlap between the price distributions, indicating that laptop type alone does not fully explain the variation in price. Other features, such as RAM, CPU, GPU, and storage, can also influence laptop prices.

##### Do different types of CPU manufacturer have different price distributions?

```python
# Lets observ the next one which is CPU manufacurer
# Do different types of CPU manufacturer have different price distributions?

fig, ax = plt.subplots(figsize=(12, 6))

df.boxplot(column="Price (Euro)", by= "CPU_Company", ax = ax)

ax.set_xlabel("CPU manufacturer")
ax.set_ylabel("Price (Euro)")
ax.set_title("Laptop Price by CPU manufacurer")

plt.suptitle("")
plt.xticks(rotation=30)

plt.show()
```


The box plot shows differences in laptop price distributions across CPU manufacturers. Intel laptops have a higher median price compared with AMD laptops. However, the Samsung category contains only one observation, so its price distribution cannot be meaningfully compared with the other manufacturers.

##### Do different types of GPU manufacturer have different price distributions?

```python
# Lets observ the next one which is GPU manufacurer
# Do different types of CPU manufacturer have different price distributions?

fig, ax = plt.subplots(figsize=(12, 6))

df.boxplot(column="Price (Euro)", by= "GPU_Company", ax = ax)

ax.set_xlabel("GPU manufacturer")
ax.set_ylabel("Price (Euro)")
ax.set_title("Laptop Price by GPU manufacurer")

plt.suptitle("")
plt.xticks(rotation=30)

plt.show()
```

As can be seen from the box plot, NVIDIA laptops have the highest mean price among the GPU manufacturers, followed by Intel and AMD. The median price of NVIDIA laptops is also higher than the median prices of Intel and AMD laptops, indicating that the price distribution of NVIDIA laptops tends to be higher. The number of laptops with an ARM GPU is very low, so there are not enough observations to draw a meaningful conclusion about its price distribution.

##### Do different types of Operating System have different price distributions?


```python
# Lets observ the next one which is Operating system
# Do different types of CPU manufacturer have different price distributions?

fig, ax = plt.subplots(figsize=(12, 6))

df.boxplot(column="Price (Euro)", by= "OpSys", ax = ax)

ax.set_xlabel("Operating System")
ax.set_ylabel("Price (Euro)")
ax.set_title("Laptop Price by Operating System")

plt.suptitle("")
plt.xticks(rotation=30)

plt.show()
```

As can be seen from the box plot, Mac and Windows 7 have relatively similar median and Q3 values, and both are higher than those of most other operating systems. Android and Chrome OS have the lowest median prices. However, the number of laptops running Android is very low, so this observation should be interpreted with caution. Mac OS X has a median price similar to the Windows operating systems, but its price distribution is narrower, indicating less variability in the middle 50% of prices.
