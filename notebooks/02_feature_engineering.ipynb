### ScreenResolution feature engineering

```python
df2["ScreenResolution"].value_counts()
```

We can extract these feature from ScreenResolution:
`Touchscreen`, `Ips`(the panel type) and `Resolution`

Let's start with the `Touchscreen`

```python
touchscreen = df2["ScreenResolution"].str.contains("Touchscreen").astype(int)
touchscreen.value_counts()
```

```python
df2.insert(5, "Touchscreen", touchscreen )
df2
```

Now we can also extract the panel type, `Ips`

```python
ips = df2["ScreenResolution"].str.contains("IPS").astype(int)
ips.value_counts()
```

```python
df2.insert(5, "IPS", ips)
df2
```

Next feature extraction will be the `Resolution`. We add tow column for that: `Res_width` and `Res_height`

```python
resolution = df2["ScreenResolution"].str.split().str[-1].str.split("x", expand=True)
res_width = resolution[0].astype(int)
res_height = resolution[1].astype(int)
print(res_width)
print(res_height)
```

```python
df2.insert(7, "Res_width", res_width)
df2.insert(8, "Res_height", res_height)
df2
```

### PPI feature engineering

Width and height alone don't show how sharp a screen is, because they ignore screen size.
The same 1920x1080 looks sharper on a 13" screen than on a 17" one. PPI (pixels per inch)
combines resolution and `Inches` into one number: higher PPI means a sharper screen,
which is usually found on more expensive laptops.

PPI = diagonal in pixels / diagonal in inches

diagonal in pixels = √(width² + height²)

```python
diagonal_in_pixels = np.sqrt((df2["Res_width"] **2) + (df2["Res_height"] **2))
PPI = diagonal_in_pixels / df2["Inches"]
PPI
```

```python
df2.insert(9, "PPI", PPI)
df2
```

### Memory feature engineering

```python
df2["Memory"].value_counts()
```

Here we can extract four feature from the Memory, since the value of the memories are different, these column should be added to our data set:

`SSD`, `HDD`, `Flash`, `Hybrid`

but before doing that, we should convert all the unit scales to one unique scale for example: GB

```python
def to_gb(size: str) -> float:
    """Convert a storage size like '128GB', '1TB' or '1.0TB' to GB."""
    if "TB" in size:
        tb = float(size.replace("TB", ""))
        return tb * 1000
    return float(size.replace("GB", ""))
    
```
