### 2. Inspecting Categroical columns

```python
df["Company"].value_counts()
```

#### Company

The `Company` feature contains several categories with highly different
frequencies. Dell, Lenovo, and HP are well represented, while several
companies have only a few observations.

We will consider how to handle these rare categories during categorical
encoding and preprocessing.

```python
df["Product"].value_counts()
```

`Product` has high cardinality (618 unique values), with many products appearing only once. This may make direct one-hot encoding inefficient, so its treatment will be considered during feature engineering.

```python
df["TypeName"].value_counts()
```

The `TypeName` contains six categories with different frequencies. The feature has relatively low cardinality and can likely be handled using one-hot encoding during preprocessing.

```python
df["ScreenResolution"].value_counts()
```

```python
df["ScreenResolution"].nunique()
```

`ScreenResolution` has relatively high cardinality, but its categorical values contain multiple meaningful attributes such as resolution, panel type, and touchscreen capability. Therefore, directly one-hot encoding the raw feature may not be the most appropriate representation. Feature extraction will be considered during preprocessing.

```python
df["CPU_Company"].value_counts()

```

`CPU_Company` has only three categories, with Intel being highly dominant. Samsung is extremely rare, appearing only once. Due to its low cardinality, the feature can be handled with one-hot encoding, while the effect of the rare Samsung category should be considered during preprocessing.
